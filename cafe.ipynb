```python
import kagglehub,pandas as pd,os

path = kagglehub.dataset_download("ahmedmohamed2003/cafe-sales-dirty-data-for-cleaning-training")
print(os.listdir(path))
df=pd.read_csv(path+'/dirty_cafe_sales.csv')
```

```python
import numpy as np

for col in df.columns:
    df[col] = df[col].replace(r'(?i)^(error|unknown)$', np.nan, regex=True)
```

```python
cols_to_float = ['Price Per Unit', 'Total Spent']
df[cols_to_float] = df[cols_to_float].apply(lambda x: pd.to_numeric(x, errors='coerce'))
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce').astype('Int64')
```
